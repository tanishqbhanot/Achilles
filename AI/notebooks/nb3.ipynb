```python
import requests

repo_url = "https://github.com/Tejasisnothere/Artemis"

parts = repo_url.rstrip("/").split("/")
owner = parts[-2]
repo = parts[-1]

url = f"https://api.github.com/repos/{owner}/{repo}/readme"

response = requests.get(url)
response.raise_for_status()

data = response.json()

readme = requests.get(data["download_url"]).text

print(readme)
```

```python
import requests


url = f"https://api.github.com/repos/{owner}/{repo}/git/trees/main?recursive=1"

response = requests.get(url)
response.raise_for_status()

tree = response.json()["tree"]
paths = []
for item in tree:
    print(item["path"], item["type"])
    paths.append(item['path'])
```

```python
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser

import os
llm = ChatGroq(model="openai/gpt-oss-120b", api_key=os.getenv("GROQ_API_KEY"))
```

```python
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="BAAI/bge-small-en-v1.5"
)

vectorstore = Chroma(
    collection_name="github_ast",
    embedding_function=embeddings,
    persist_directory="./chroma_db"
)

```

```python
from langchain_core.documents import Document
```

```python
documents = paths

vectorstore.add_texts(
    texts=documents
)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = PromptTemplate.from_template("""
User has the following skills:
{skills}

Based on the following README:
{readme}

Generate one question relevant to what the user worked on, using their skills to test their understanding.

Return only the question.
""")

chain = prompt | llm | StrOutputParser()

question = chain.invoke({
    "skills": skills,
    "readme": readme
})

print(question)
```

```python
query = f"""
Skills: {skills}

Question: {question}
"""

docs = vectorstore.similarity_search(
    query,
    k=10
)

extracted_paths = [doc.page_content for doc in docs]

print(extracted_paths)
```

```python
from pydantic import BaseModel, Field

class RelevantPaths(BaseModel):
    paths: list[str] = Field(
        description="Relevant repository paths"
    )
prompt = PromptTemplate.from_template("""
You are selecting relevant code paths from a GitHub repository.

User's skills:
{skills}


Available GitHub AST paths:
{paths}

Task:
Select only the paths relevant to the user's skills and the interview question.

Rules:
- Only select paths from the provided list.
- Do not invent paths.
- Return an empty list if no paths are relevant.
- Return ONLY a valid JSON object.
- The JSON must follow this exact format:

{{
    "paths": ["path1.py", "path2.py"]
}}
""")
structured_llm = llm.with_structured_output(
    RelevantPaths,
    method="json_mode"
)
chain = prompt | structured_llm

result = chain.invoke({
    "skills": skills,
    "paths": paths
})

print(result.paths)
```
